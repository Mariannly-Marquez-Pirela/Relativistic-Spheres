# Martching Conditions: Spherical Star & Compact Object 

Simple Boundary Surface


Interior: Perfect fluid (Tolman IV)

Exterior: Empty space (Schwarzschild)

### Coordinates and Function to Compute Scalars

```python
import sympy as sp

# ============================================
# Coordinates
# ============================================

r, t, theta = sp.symbols('r t theta', real=True)

# ============================================
# Function
# ============================================

def compute_variables(A, B, R):

    a1 = sp.diff(A, r)/(B*A)

    sigma1 = sp.diff(B, t)/(A*B)

    sigma2 = sp.diff(R, t)/(A*R)

    J1 = sp.diff(R, r)/(B*R)

    J2 = sp.cot(theta)/R

    return a1, sigma1, sigma2, J1, J2
```

### Schwarzschild Metric Functions

```python
m = sp.Symbol('m', positive=True)

A1 = sp.sqrt(1-2*m/r)
B1 = 1/ sp.sqrt(1-2*m/r)
R1 = r

a1_Schwarz, sigma1_Schwarz, sigma2_Schwarz, J1_Schwarz, J2_Schwarz = compute_variables(A1, B1, R1)

print("\n=== Schwarzschild ===")
print("a1= ", a1_Schwarz)
print("sigma1= ", sigma1_Schwarz)
print("sigma2= ", sigma2_Schwarz)
print("J1= ", J1_Schwarz)
print("J2= ", J2_Schwarz)
```

```python
A1_latex = sp.latex(A1)
print(A1_latex)
```

```python
B1_latex = sp.latex(B1)
print(B1_latex)
```

### Tolman IV Metric Functions

```python
H, J, K = sp.symbols('H J K', positive=True)

A2 = sp.sqrt(K**2 * (1 + r**2/J**2))
#B2 = sp.sqrt((1 + 2*r**2/J**2) / ((1 - r**2/H**2)*(1 + r**2/J**2)))
B2_sq = (1 + 2*r**2/J**2) / ((1 - r**2/H**2)*(1 + r**2/J**2))
B2 = sp.sqrt(B2_sq)
R2 = r
P_minus = (K**2*(1 - r**2/H**2) - (1 - 2*m/r)) / (8*sp.pi*r**2)

a1_Tol, sigma1_Tol, sigma2_Tol, J1_Tol, J2_Tol = compute_variables(A2, B2, R2)

print("\n=== Tolman IV ===")
print("a1= ", a1_Tol)
print("sigma1= ", sigma1_Tol)
print("sigma2= ", sigma2_Tol)
print("J1= ", J1_Tol)
print("J2= ", J2_Tol)
```

### Continuity of First Fundamental Form

$$\dot c= \sqrt{ \frac{\llbracket A^2 \rrbracket}{\llbracket B^2 \rrbracket } }$$


```python
c0=sp.sqrt((A1**2-A2**2)/(B1**2-B2**2))
print(c0)
```

```python
latex = sp.latex(c0)
print(latex)
```

### Israel Condition

$$\dot{c} = \frac{\llbracket A (\sigma_1 - a_1 )\rrbracket}{\llbracket B (\sigma_1 - a_1 ) \rrbracket}$$
